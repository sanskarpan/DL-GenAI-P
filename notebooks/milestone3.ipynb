```python
import torch
import torch.nn as nn

x = torch.randn(32, 1, 64, 64)
flat = x.flatten(start_dim=1)  # flatten channels * height * width
print(f"Original shape: {x.shape}")
print(f"Flattened shape: {flat.shape}")
print(f"Input features for Linear layer: {flat.shape[1]}")
```

```python
import torchaudio

waveform = torch.randn(1, 16000)  # 1 channel, 16000 samples (1 second)
mel_transform = torchaudio.transforms.MelSpectrogram(
    sample_rate=16000, n_fft=400, hop_length=160, n_mels=64
)
mel_spec = mel_transform(waveform)
print(f"Waveform shape: {waveform.shape}")
print(f"MelSpectrogram shape: {mel_spec.shape}")
```

```python
conv = nn.Conv2d(1, 16, kernel_size=3, stride=1, padding=1)
pool = nn.MaxPool2d(kernel_size=2, stride=2)

x = torch.randn(1, 1, 64, 64)
after_conv = conv(x)
after_pool = pool(after_conv)

print(f"Input:         {x.shape}")           
print(f"After Conv2d:  {after_conv.shape}")  
print(f"After MaxPool: {after_pool.shape}")  
print(f"Spatial output: ({after_pool.shape[2]}, {after_pool.shape[3]})")
```

```python

layer = nn.Linear(in_features=128, out_features=64)
n_weights = 128 * 64
n_biases = 64
total = sum(p.numel() for p in layer.parameters())

print(f"Weight matrix: {layer.weight.shape} = {n_weights} params")
print(f"Bias vector:   {layer.bias.shape} = {n_biases} params")
print(f"Total learnable parameters: {total}")
```

```python
import math
from torch.utils.data import DataLoader, TensorDataset

dataset = TensorDataset(torch.randn(1050, 1))
loader = DataLoader(dataset, batch_size=32, drop_last=False)

print(f"Dataset size: {len(dataset)}")
print(f"Batch size: 32")
print(f"ceil(1050 / 32) = ceil({1050/32:.4f}) = {math.ceil(1050/32)}")
print(f"Actual batches in DataLoader: {len(loader)}")
```

```python
logits = torch.tensor([[2.5, 1.0, 0.1]])
target = torch.tensor([0])
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(logits, target)

# Manual computation:
# softmax = exp(logits) / sum(exp(logits))
import numpy as np
exp_logits = np.exp([2.5, 1.0, 0.1])
softmax = exp_logits / exp_logits.sum()
manual_loss = -np.log(softmax[0])

print(f"Logits: [2.5, 1.0, 0.1], Target: class 0")
print(f"Softmax probabilities: {softmax}")
print(f"P(class 0) = {softmax[0]:.4f}")
print(f"Loss = -log({softmax[0]:.4f}) = {manual_loss:.6f}")
print(f"PyTorch CrossEntropyLoss: {loss.item():.3f}")
```

```python
# Formula: new_weight = weight - lr * gradient
weight = torch.tensor([0.5], requires_grad=False)
gradient = 0.2
lr = 0.01

new_weight = 0.5 - 0.01 * 0.2
print(f"Current weight: 0.5")
print(f"Gradient: 0.2")
print(f"Learning rate: 0.01")
print(f"new_weight = 0.5 - 0.01 * 0.2 = {new_weight}")

w = torch.tensor([0.5], requires_grad=True)
loss = w * 1.0  # dummy loss so grad = d(loss)/d(w)
loss.backward()   
w.grad = torch.tensor([0.2]) 
optimizer = torch.optim.SGD([w], lr=0.01)
optimizer.step()
print(f"PyTorch SGD result: {w.item()}")
```

```python
# Formula: padding = (kernel_size - 1) / 2 = (5 - 1) / 2 = 2
padding = (5 - 1) // 2
print(f"kernel_size=5, stride=1")
print(f"padding = (5 - 1) // 2 = {padding}")

conv = nn.Conv2d(1, 1, kernel_size=5, stride=1, padding=padding)
x = torch.randn(1, 1, 64, 64)
out = conv(x)
print(f"Input:  {x.shape}")
print(f"Output: {out.shape}")
print(f"Same size: {x.shape[2:] == out.shape[2:]}")
```
